```python
import json
import numpy as np
import re
import pymorphy3
import pandas as pd
```

```python
with open("result.json", "r", encoding="utf8") as file:
    data = json.load(file)

messages = data["messages"]

print(len(messages))
```

```python
messages
```

```python
texts = []
for i in messages:
    if i.get("from") == "ydhshss":
        text = i.get("text")

        if isinstance(text,str):
            texts.append(text)

texts
```

```python
full_text = " ".join(texts)

tokens = re.findall(r"\b\w+\b", full_text.lower())
print("Количество токенов:", len(tokens))
```

```python
full_text[:50]
```

```python
morph = pymorphy3.MorphAnalyzer()

lemas = []

for token in tokens:
    lemma = morph.parse(token)[0].normal_form
    lemas.append(lemma)

print(lemas[:50])
```

```python
unique_lemas = set(lemas)
print(len(unique_lemas))
```
